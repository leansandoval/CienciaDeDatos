## **Ejercicio 1**

Una vez que el sistema de recaudaciones del nuevo canal de cobranzas está en producción se recibieron las siguientes transacciones correspondientes a pagos en pesos: 100, 300, 400, 500, 6000, 8000, 2012, 1231, 23, 20, 100, 100, 4000, 300, 500, 500, 600, 200, 200, 8000.

```python
pagos = [100, 300, 400, 500, 6000, 8000, 2012, 1231, 23, 20, 100, 100, 4000, 300, 500, 500, 600, 200, 200, 8000]
```

---
### Ejercicio 1.1

Genere una variable con la cantidad de pagos recibidos denominada “cant_pagos”.

```python
cant_pagos = len(pagos)
print(f"Cantidad de pagos recibidos: {cant_pagos}")
```

**Descripción:** Se calcula la cantidad de pagos recibidos utilizando la función `len()` sobre la lista de pagos.


---
### Ejercicio 1.2

Genere un vector con los valores únicos de los pagos realizados llamado "valores_unicos".

```python
valores_unicos = list(set(pagos))
print(f"Valores únicos de pagos: {valores_unicos}")
```

**Descripción:** Se utiliza `set()` para obtener los valores únicos de la lista de pagos y luego se convierte a lista para su visualización.


---
### Ejercicio 1.3

Ordene de menor a mayor los pagos en un vector llamado "pagos_ordenados".

```python
pagos_ordenados = sorted(pagos)
print(f"Pagos ordenados: {pagos_ordenados}")
```

**Descripción:** Se utiliza la función `sorted()` para ordenar la lista de pagos de menor a mayor.


---
### Ejercicio 1.4

¿Cuál es el mínimo valor que se cobró y cual es el máximo valor que se cobró?

```python
minimo = min(pagos)
maximo = max(pagos)
print(f"Valor mínimo cobrado: {minimo}")
print(f"Valor máximo cobrado: {maximo}")
```

**Descripción:** Se utilizan las funciones `min()` y `max()` para obtener el valor mínimo y máximo de la lista de pagos.


---
### Ejercicio 1.5

Genere un nuevo vector denominado "comision" con las comisiones de las cobranzas que se deben abonar al recaudador, recuerde que se cobra el 1.2%.

```python
comision = [round(p * 0.012, 2) for p in pagos]
print(f"Comisiones calculadas: {comision}")
```

**Descripción:** Se calcula la comisión para cada pago multiplicando por 0.012 y redondeando a dos decimales.


---
### Ejercicio 1.6

Realice otro vector denominado "comision_pagada" teniendo en cuenta que si la comisión no supera los 8 pesos el valor mínimo de cobranzas es de 8 pesos.

```python
comision_pagada = [c if c >= 8 else 8 for c in comision]
print(f"Comisión pagada (mínimo 8): {comision_pagada}")
```

**Descripción:** Se aplica la condición para que la comisión mínima pagada sea de 8 pesos, usando una lista por comprensión.

**Descripción:** Se cuenta cuántas comisiones pagadas son exactamente 8 pesos usando el método `count()`.

### Ejercicio 2.1

Generar la función denominada calculoIMC que tenga como parámetros peso y altura. Tener en cuenta:

*   Convertir la Altura de pulgadas a Metros (indice 0.0254)
*   Convertir el Peso de libras a KG (índice 0.453592)
*   La función es Peso dividido la Altura al cuadrado.

```python
INDICE_ALTURA = 0.0254
INDICE_PESO = 0.453592
def calculoIMC(peso_libras, altura_pulgadas):
    peso_kg = peso_libras * INDICE_PESO
    altura_m = altura_pulgadas * INDICE_ALTURA
    imc = peso_kg / (altura_m ** 2)
    return imc
```

Verificar que la función para el siguiente caso:

```python
peso_test = 112.9925
altura_test = 65.78331
imc_resultado = calculoIMC(peso_test, altura_test)
print(f"IMC para peso {peso_test} libras y altura {altura_test} pulgadas: {imc_resultado:.2f}")
```

## **Ejercicio 3**

Un sistema registró los siguientes datos de clientes: edades, ingresos mensuales en dólares, y géneros.

```python
import pandas as pd

data = {
    "edad": [25, 34, 22, 45, 52, 23, 31, 40, 29, 37, 28],
    "ingreso": [1800, 2500, 1200, 4000, 5200, 1300, 2100, 3000, 1600, 2700, 2000],
    "genero": ["F", "M", "F", "M", "M", "F", "F", "M", "F", "M", "F"]
}
```

---
### Ejercicio 3.1

¿Cuál es la edad promedio de los clientes?

```python
df = pd.DataFrame(data)
print("Datos de clientes:")
print(df)
```

```python
edad_promedio = df['edad'].mean()
print(f"Edad promedio de los clientes: {edad_promedio:.2f} años")
```

---
### Ejercicio 3.2

¿Cuál es el ingreso máximo y el mínimo?

```python
ingreso_maximo = df['ingreso'].max()
ingreso_minimo = df['ingreso'].min()
print(f"Ingreso máximo: ${ingreso_maximo}")
print(f"Ingreso mínimo: ${ingreso_minimo}")
```

---
### Ejercicio 3.3

Calcule el ingreso promedio para mujeres y para varones.

```python
ingreso_por_genero = df.groupby('genero')['ingreso'].mean()
print("Ingreso promedio por género")
print(f"Mujeres (F): ${ingreso_por_genero['F']:.2f}")
print(f"Varones (M): ${ingreso_por_genero['M']:.2f}")
```

---
### Ejercicio 3.4

Genere una lista de booleanos que indique si cada cliente gana más que el promedio general.

```python
ingreso_promedio_general = df['ingreso'].mean()
gana_mas_promedio = df['ingreso'] > ingreso_promedio_general
print(f"Ingreso promedio general: ${ingreso_promedio_general:.2f}")
```

```python
print("¿Gana más que el promedio?")
for i, (cliente, gana_mas) in enumerate(zip(df['ingreso'], gana_mas_promedio)):
    print(f"Cliente {i+1} (${cliente}): {gana_mas}")
```

```python
lista_booleanos = gana_mas_promedio.tolist()
print(f"Lista de booleanos: {lista_booleanos}")
```
