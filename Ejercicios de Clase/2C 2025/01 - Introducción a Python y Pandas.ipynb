### **Ejercicio 1**

Una vez que el sistema de recaudaciones del nuevo canal de cobranzas está en producción se recibieron las siguientes transacciones correspondientes a pagos en pesos: 100, 300, 400, 500, 6000, 8000, 2012, 1231, 23, 20, 100, 100, 4000, 300, 500, 500, 600, 200, 200, 8000.

1. Genere una variable con la cantidad de pago recibidos denominada “cant_pagos”.
2. Genere un vector con los valores únicos de los pagos realizados llamado "valores_unicos".
3. Ordene de menor a mayor los pagos en un vector llamado "pagos_ordenados".
4. ¿Cuál es el mínimo valor que se cobró y cual es el máximo valor que se cobró?
5. Genere un nuevo vector denominado "comision" con las comisiones de las cobranzas que se deben abonar al recaudador, recuerde que se cobra el 1.2%.
6. Realice otro vector denominado "comision_pagada" teniendo en cuenta que si la comisión no supera los 8 pesos el valor mínimo de cobranzas es de 8 pesos.
7. Cuántas operaciones se pagaron 8 pesos por no superar ese tope?
8. ¿Cuánto pagué de comisiones y cuanto deberia haber pagado sin no estuviera el tope?

```python
pagos = [100, 300, 400, 500, 6000, 8000, 2012, 1231, 23, 20, 100, 100, 4000, 300, 500, 500, 600, 200, 200, 8000]
```

---
#### Resolución Ejercicio 1
A continuación se resuelven los incisos solicitados utilizando listas y funciones básicas de Python.


```python
# 1. Cantidad de pagos recibidos
cant_pagos = len(pagos)
print(f"Cantidad de pagos recibidos: {cant_pagos}")
```

**Descripción:** Se calcula la cantidad de pagos recibidos utilizando la función `len()` sobre la lista de pagos.


```python
# 2. Vector con valores únicos de pagos realizados
valores_unicos = list(set(pagos))
print(f"Valores únicos de pagos: {valores_unicos}")
```

**Descripción:** Se utiliza `set()` para obtener los valores únicos de la lista de pagos y luego se convierte a lista para su visualización.


```python
# 3. Ordenar pagos de menor a mayor
pagos_ordenados = sorted(pagos)
print(f"Pagos ordenados: {pagos_ordenados}")
```

**Descripción:** Se utiliza la función `sorted()` para ordenar la lista de pagos de menor a mayor.


```python
# 4. Mínimo y máximo valor cobrado
minimo = min(pagos)
maximo = max(pagos)
print(f"Valor mínimo cobrado: {minimo}")
print(f"Valor máximo cobrado: {maximo}")
```

**Descripción:** Se utilizan las funciones `min()` y `max()` para obtener el valor mínimo y máximo de la lista de pagos.


```python
# 5. Vector de comisiones (1.2%)
comision = [round(p * 0.012, 2) for p in pagos]
print(f"Comisiones calculadas: {comision}")
```

**Descripción:** Se calcula la comisión para cada pago multiplicando por 0.012 y redondeando a dos decimales.


```python
# 6. Vector de comisión pagada (mínimo 8 pesos)
comision_pagada = [c if c >= 8 else 8 for c in comision]
print(f"Comisión pagada (mínimo 8): {comision_pagada}")
```

**Descripción:** Se aplica la condición para que la comisión mínima pagada sea de 8 pesos, usando una lista por comprensión.


```python
# 7. Operaciones que pagaron 8 pesos por el tope
cant_8pesos = comision_pagada.count(8)
print(f"Cantidad de operaciones con comisión mínima de 8 pesos: {cant_8pesos}")
```

**Descripción:** Se cuenta cuántas comisiones pagadas son exactamente 8 pesos usando el método `count()`.


```python
# 8. Total pagado de comisiones y total sin tope
total_comision_pagada = sum(comision_pagada)
total_comision_sin_tope = sum(comision)
print(f"Total pagado de comisiones (con tope): {total_comision_pagada}")
print(f"Total que se hubiera pagado sin tope: {total_comision_sin_tope}")
```

**Descripción:** Se suman los valores de ambos vectores para comparar el total pagado de comisiones con y sin el tope mínimo.
